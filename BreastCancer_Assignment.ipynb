```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score, confusion_matrix)
from sklearn.model_selection import cross_val_score

#model used Decision tree classifier

url = pd.read_csv("data.csv")
print("dataset before dropping id column:\n")
print(url.head())

if 'id' in url.columns:
    url.drop(columns=['id'], inplace=True)

if 'Unnamed: 32' in url.columns:
    url.drop(columns=['Unnamed: 32'], inplace=True)
df = pd.DataFrame(url)

print("dataset after dropping id and Unnamed: 32 column:\n")
print(url.head())


#categorical columns
print("categorical columns")
print(df.select_dtypes(include = 'object').columns.tolist())

#Numerical columns
print("\n Numerical columns")
print(df.select_dtypes(include = np.number).columns.tolist())

#checking null values in the dataframe
print("\n Missing values")
missing = df.isnull().sum()
print(missing)
missing_percent = (missing / len(df))* 100
missing_table = pd.DataFrame({
    "Missing Count": missing,
    "Missing Percentage" : missing_percent
})
print(missing_table.head())

#since there is no missing values - proceed with converting categorical target data to numerical values

df['diagnosis'] = df['diagnosis'].map({'M':1, 'B':0})

x = df.drop('diagnosis', axis=1)
y = df['diagnosis']

#test - size = 0.3 means 30% of the data is passed for testing remaining for training
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42)

#training the model with training dataset
model = DecisionTreeClassifier()
model.fit(x_train, y_train)

#testing the model with training dataset
print("\n Prediction on training data\n")
pred_x = model.predict(x_train)
print("\n",pred_x)

#testing the model with 30% testing dataset

print("\n Prediction on testing data\n")
pred_y = model.predict(x_test)
print("\n",pred_y)

#evalution part
print("\n Evaluating the training results\n")
print("Accuracy : ", accuracy_score(y_train, pred_x))
print("Precision : ", precision_score(y_train, pred_x, average='weighted'))
print("Recall : ", recall_score(y_train, pred_x, average='weighted'))
print("F1 Score : ", f1_score(y_train, pred_x, average='weighted'))

print("\n Evaluating the testing results\n")
print("Accuracy : ", accuracy_score(y_test, pred_y))
print("Precision : ", precision_score(y_test, pred_y, average='weighted'))
print("Recall : ", recall_score(y_test, pred_y, average='weighted'))
print("F1 Score : ", f1_score(y_test, pred_y, average='weighted'))

#Confusion matrix
conf_matrix = confusion_matrix(y_test, pred_y)
print("\n Confusion Matrix : \n", conf_matrix)

#K-fold cross validation - here K = 5(1 part of the dataset to test and 4 parts to train)
print("\n 5 fold cross validation \n")
cv_scores = cross_val_score(model, x, y, cv=5, scoring = 'accuracy')
print("CV scores: " ,cv_scores)
print("Mean CV Accuarcy: ", cv_scores.mean())












```
